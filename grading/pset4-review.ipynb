```python
import sys
from pathlib import Path

# Run this notebook from stringreps/grading/.
root = Path.cwd().resolve().parent
src = root / "pset-4" / "src"
if not (src / "pset4").is_dir():
    raise FileNotFoundError(
        "Run this notebook from stringreps/grading/; expected ../pset-4/src/pset4."
    )
sys.path.insert(0, str(src))

# Clear cached imports so rerunning this cell loads this submission.
for name in list(sys.modules):
    if name == "pset4" or name.startswith("pset4."):
        del sys.modules[name]

from pset4 import (
    hammingDistance,
    approximatePatternMatching,
    frequentWordsWithMismatches,
    frequentWordsWithMismatchesAndReverseComplements,
)
```

```python
# Test Hamming Distance
# one edit
sequence1 = "AACCTTGG"
sequence2 = "AACCTGGG"
distance = hammingDistance(sequence1, sequence2)
assert distance == 1

# same string
sequence2 = "AACCTTGG"
distance = hammingDistance(sequence1, sequence2)
assert distance == 0

# reverse complement
sequence2 = "CCAAGGTT"
distance = hammingDistance(sequence1, sequence2)
assert distance == 8
```

```python
# Test Approximate Pattern Matching

# Single-nucleotide pattern matching
sequence = "AAAAAAAAAA"
output = approximatePatternMatching("A", sequence, 1)
assert output == list(range(10))

# Single-nucleotide pattern matching with errors
output = approximatePatternMatching("G", sequence, 1)
assert output == list(range(10))

# 2-mer pattern matching
output = approximatePatternMatching("AA", sequence, 1)
assert output == list(range(9))

# 2-mer pattern with errors
output = approximatePatternMatching("AG", sequence, 1)
assert output == list(range(9))

# Pattern is entire sequence
output = approximatePatternMatching("AAAAAAAAAA", sequence, 1)
assert output == [0]

# Pattern longer than sequence
output = approximatePatternMatching("GGGGGGGGGGGGGGGGGGGGG", sequence, 1)
assert output == []
```

```python
# Test Frequent Words with Mismatches

# Zero mismatches for a 2-mer
text = "AAAAAAAAA"
k = 2
d = 0
assert frequentWordsWithMismatches(text, k, d)  == {'AA'}

# One position in 2-mer can deviate -- checks values
d = 1
assert set(frequentWordsWithMismatches(text, k, d)) == {'AA', 'AC', 'AG', 'AT', 'CA', 'GA', 'TA'}

# Both positions in 2-mer can deviate -- checks length
d = 2
assert len(frequentWordsWithMismatches(text, k, d)) == 4**2

# One dominant motif
text = "AAAAAAAAGGGGGG"
k = 3
d = 0
assert frequentWordsWithMismatches(text, k, d) == {"AAA"}

# One dominant motif + flexible d
d = 3
assert len(frequentWordsWithMismatches(text, k, d)) == 4**3

# Two equally dominant motifs
text = "".join(["A"] * 9 + ["G"] * 9)
d = 0
assert set(frequentWordsWithMismatches(text, k, d)) == {"AAA", "GGG"}
```

```python
# Test frequentWordsWithMismatchesAndReverseComplements

# Zero mismatches for a 2-mer
text = "AAAAAAAAA"
k = 2
d = 0
assert frequentWordsWithMismatchesAndReverseComplements(text, k, d)  == {'AA', 'TT'}

# One position in 2-mer can deviate -- checks values
d = 1
assert set(frequentWordsWithMismatchesAndReverseComplements(text, k, d)) == {'AT', 'TA'}

# Both positions in 2-mer can deviate -- checks length
d = 2
assert len(frequentWordsWithMismatchesAndReverseComplements(text, k, d)) == 4**2

# One dominant motif
text = "AAAAAAAAGGGGGG"
k = 3
d = 0
assert frequentWordsWithMismatchesAndReverseComplements(text, k, d) == {"AAA", "TTT"}

# One dominant motif + flexible d
d = 3
assert len(frequentWordsWithMismatchesAndReverseComplements(text, k, d)) == 4**3

# Two equally dominant motifs
text = "".join(["A"] * 9 + ["G"] * 9)
d = 0
assert set(frequentWordsWithMismatchesAndReverseComplements(text, k, d)) == {"AAA", "TTT", "GGG", "CCC"}
```
