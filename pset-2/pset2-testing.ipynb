```python
# This is support code to help generate informative errors
class NotImplementedYet(AssertionError):
    pass


def run_function(func, *args, **kwargs):
    try:
        return func(*args, **kwargs)
    except NotImplementedError as e:
        raise NotImplementedYet(
            f"Your {func.__name__} function was found and called successfully! "
            "Now make sure you replace the starter NotImplementedError "
            "with your implementation."
        ) from e
```

## Problem 5: countNucleotides
This function expects:
    Input: text, which is a string
    Return: dictionary, with keys A,C,T,G and the frequency of each associated nucleotide in the text string

```python
sequence = "ACTGACTG"
counts = list(countNucleotides(sequence).values())
assert counts == [2] * 4

sequence = "ABRACADABRA"
try:
    counts = countNucleotides(sequence)
    assert counts.get("A") == 5, (f"For {sequence}, expected four nucleotide counts of 2, but got {counts}")
    assert counts.get("C") == 1, (f"For {sequence}, expected A count of 5, but got {counts['A']}")
    assert "B" not in counts.keys(), (f"Expected dictionary to contain only nucleotide keys, but got {counts.keys()}")
except ValueError:
    pass

sequence = "A"
counts = countNucleotides(sequence)
assert counts.get("A") == 1, (f"For {sequence}, expected A count = 1, but got {counts['A']}")
assert counts.get("G") == 0, (f"For {sequence}, expected G count = 0, but got {counts['G']}")
```

## Problem 6: patternIndex

```python
sequence = "GGGG"

result = patternIndex(sequence, "G")
assert result == [0, 1, 2, 3], (
    f"Expected patternIndex('{sequence}', 'G') to return [0, 1, 2, 3], "
    f"but got {result}. Check that you are returning every matching index."
)

result = patternIndex(sequence, "C")
assert result == [], (
    f"Expected patternIndex('{sequence}', 'C') to return [], "
    f"but got {result}. A pattern that does not occur should return an empty list."
)

sequence = "GCTG"
result = patternIndex(sequence, "G")
assert result == [0, 3], (
    f"Expected patternIndex('{sequence}', 'G') to return [0, 3], "
    f"but got {result}. Check that you are examining the entire sequence, including the final position."
)

sequence = "TCTCTCTC"

result = patternIndex(sequence, "TCT")
assert result == [0, 2, 4], (
    f"Expected patternIndex('{sequence}', 'TCT') to return [0, 2, 4], "
    f"but got {result}. Check that your implementation finds overlapping pattern occurrences."
)

result = patternIndex(sequence, "CTC")
assert result == [1, 3, 5], (
    f"Expected patternIndex('{sequence}', 'CTC') to return [1, 3, 5], "
    f"but got {result}. Check overlapping matches and starting indices."
)

sequence = "GCTG"
result = patternIndex(sequence, "GCTG")
assert result == [0], (
    f"Expected patternIndex('{sequence}', 'GCTG') to return [0], "
    f"but got {result}. A pattern equal to the full sequence should be found at index 0."
)
```

## Problem 7: patternCount
These are the same tests from pset-1.

```python
# Test: clearly differentiated k-mers
result = run_function(patternCount, "ATAT", "AT")
assert result == 2, (
    f'Expected 2 for PatternCount("ATAT", "AT"), got {result!r}.'
)

# Test: overlapping windows matter
result = run_function(patternCount, "GCGCG", "GCG")
assert result == 2, (
    f'Expected 2 overlapping occurrences of "GCG" in "GCGCG", '
    f"got {result!r}."
)

# Test: full text
result = run_function(patternCount, "ACGT", "ACGT")
assert result == 1, (
    f"A pattern equal to the full text should occur once; got {result!r}."
)

# Test: pattern longer than text
try:
    result = run_function(patternCount, "ACGT", "AAAAA")
except NotImplementedYet:
    raise
except Exception as e:
    print(
        f"PatternCount raised {type(e).__name__} when the pattern "
        "was longer than the text. This is an acceptable way to handle this case."
    )
else:
    assert result == 0, (
        f"When the pattern is longer than the text, either return 0 "
        f"or raise an error; got {result!r}."
    )


# Test: empty text
try:
    result = run_function(patternCount, "", "A")
except NotImplementedYet:
    raise
except Exception as e:
    print(
        f"PatternCount raised {type(e).__name__} for empty text. "
        "This is another acceptable way to handle this case."
    )
else:
    assert result == 0, (
        f"An empty text should either return 0 or raise an error; "
        f"got {result!r}."
    )
```

## Problem 8: Frequent Words
These are the same tests from pset-1

```python
# Test: one clear winner 
result = run_function(
    frequentWords,
    "ATAT",
    2
)

assert result == {"AT"}, (
    f'Expected {{"AT"}}, got {result!r}. '
    '"AT" occurs twice in "ATAT".'
)

# Test: one clear winner with overlapping windows
result = run_function(
    frequentWords,
    "ATATATACGTCGT",
    3
)

assert result == {"ATA"}, (
    f'Expected {{"ATA"}}, got {result!r}. '
    '"ATA" occurs three times because overlaps count.'
)

# Test: two-way tie
result = run_function(
    frequentWords,
    "TCGACGGAT",
    2
)

assert result == {"GA", "CG"}, (
    f'Expected {{"GA", "CG"}}, got {result!r}. '
    "More than one k-mer can have the maximum count. "
    "Make sure you return every k-mer tied for most frequent."
)


# Test: final window matters
result = run_function(
    frequentWords,
    "ACTGACTGACTGAC",
    3
)

assert result == {"ACT", "CTG", "TGA", "GAC"}, (
    f'Expected {{"ACT", "CTG", "TGA", "GAC"}}, got {result!r}. '
    "Check that you examine every possible length-k window, including "
    "the final window, and include every tie for the maximum count."
)
```
